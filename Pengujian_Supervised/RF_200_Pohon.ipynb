# RANDOM FOREST - 200 POHON (GINI)

### Deskripsi Teori & Varian:
* **Algoritma**: Random Forest (Gini)
* **Varian**: Jumlah Pohon Lebih Banyak (n_estimators = 200)
* **Konsep**: Meningkatkan jumlah pohon keputusan dari 100 menjadi 200. Semakin banyak pohon, semakin stabil dan akurat hasil *voting* (suara mayoritas) karena mengurangi varians (noise). Namun, konsekuensinya adalah waktu komputasi yang lebih lama.

```python
import time
import tracemalloc
import re
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

df_hoaks = pd.read_csv('../data/komdigi_hoaks.csv')
df_valid = pd.read_csv('../web/uploads/berita_valid_terbaru.csv')

def clean_text(text):
    if not isinstance(text, str): return ''
    text = re.sub(r'\[(HOAKS|DISINFORMASI|MISINFORMASI|KLARIFIKASI|SALAH|FALSE)\]', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\((HOAKS|DISINFORMASI|MISINFORMASI|KLARIFIKASI|SALAH|FALSE)\)', '', text, flags=re.IGNORECASE)
    text = re.sub(r'https?:\/\/\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    return re.sub(r'\s+', ' ', text.lower()).strip()

df_hoaks['teks_bersih'] = (df_hoaks['title'].fillna('') + ' ' + df_hoaks['excerpt'].fillna('')).apply(clean_text)
df_valid['teks_bersih'] = (df_valid['title'].fillna('') + ' ' + df_valid['excerpt'].fillna('')).apply(clean_text)

df_hoaks['label'] = 1
df_valid['label'] = 0

df_hoaks_sampel = df_hoaks.sample(n=len(df_valid), random_state=42)
df_all = pd.concat([df_hoaks_sampel, df_valid], ignore_index=True)
df_all = df_all[df_all['teks_bersih'].str.len() > 10].reset_index(drop=True)

X_train, X_test, y_train, y_test = train_test_split(
    df_all['teks_bersih'], df_all['label'], test_size=0.2, random_state=42, stratify=df_all['label']
)
tfidf = TfidfVectorizer(max_features=1500, ngram_range=(1, 2))
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

print("Preprocessing Selesai!")
```

```python
tracemalloc.start()
waktu_mulai = time.time()

# Model Random Forest - 200 Pohon
model = RandomForestClassifier(n_estimators=200, criterion='gini', max_depth=None, random_state=42)
model.fit(X_train_vec, y_train)
y_pred = model.predict(X_test_vec)

waktu_eksekusi = time.time() - waktu_mulai
current_mem, peak_mem = tracemalloc.get_traced_memory()
tracemalloc.stop()
memori_mb = peak_mem / (1024 * 1024)

acc = accuracy_score(y_test, y_pred)
prec_macro = precision_score(y_test, y_pred, average='macro', zero_division=0)
rec_macro = recall_score(y_test, y_pred, average='macro', zero_division=0)
f1_macro = f1_score(y_test, y_pred, average='macro', zero_division=0)
f1_weighted = f1_score(y_test, y_pred, average='weighted', zero_division=0)

print("=============================================")
print("HASIL PENGUJIAN: RANDOM FOREST - 200 POHON")
print("=============================================")
print(f"1. Algoritma        : Random Forest")
print(f"2. Metrik Jarak     : Gini Impurity (200 Pohon)")
print(f"3. Akurasi          : {acc:.4f}")
print(f"4. Precision (Macro): {prec_macro:.4f}")
print(f"5. Recall (Macro)   : {rec_macro:.4f}")
print(f"6. F1-Score (Macro) : {f1_macro:.4f}")
print(f"7. F1-Score (Weighted): {f1_weighted:.4f}")
print(f"8. Waktu Eksekusi   : {waktu_eksekusi:.4f} detik")
print(f"9. Penggunaan Memori: {memori_mb:.2f} MB")
print("=============================================\n")

print("Laporan Klasifikasi Detail:")
print(classification_report(y_test, y_pred, target_names=['Valid (0)', 'Hoaks (1)'], zero_division=0))
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Prediksi Valid', 'Prediksi Hoaks'],
            yticklabels=['Aktual Valid', 'Aktual Hoaks'])
plt.title('Confusion Matrix: RANDOM FOREST - 200 POHON')
plt.ylabel('Label Sebenarnya')
plt.xlabel('Label Prediksi')
plt.tight_layout()
plt.show()
```

### Pembahasan Akademis: Random Forest (200 Pohon)

1. **Analisis Performa & Akurasi**:
   * Model Random Forest dengan 200 pohon menghasilkan akurasi sebesar **91.00%** dengan F1-Score 0.9099.
   * Menariknya, akurasi ini justru **sedikit lebih rendah** dibandingkan model 100 pohon (91.33%). Hal ini membuktikan bahwa penambahan jumlah pohon dari 100 ke 200 **tidak selalu meningkatkan performa** model. Pada titik tertentu, model sudah mencapai konvergensi dan penambahan pohon hanya akan menambah redundansi (pohon-pohon baru memberikan suara yang sama dengan pohon sebelumnya).

2. **Trade-off Komputasi (Waktu vs Akurasi)**:
   * Waktu eksekusi melonjak drastis menjadi **2.5361 detik**, lebih dari dua kali lipat dibanding 100 pohon (0.9907 detik). 
   * Hal ini sangat wajar karena Random Forest harus melatih 200 pohon secara berurutan, yang memakan waktu komputasi dua kali lebih banyak. Dalam konteks *deployment* sistem *real-time*, penambahan 100 pohon ini tidak sepadan dengan penurunan akurasi yang terjadi.

3. **Kesimpulan Hyperparameter Tuning**:
   * Eksperimen ini membuktikan pentingnya *Hyperparameter Tuning*. Nilai `n_estimators=100` terbukti lebih optimal (akurasi lebih tinggi, waktu lebih cepat) dibandingkan `n_estimators=200` untuk dataset hoaks ini. Menambah kompleksitas model tidak selalu berbanding lurus dengan peningkatan kualitas.
